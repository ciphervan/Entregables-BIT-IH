## 1. Distribución del tiempo diario de uso de redes sociales

Se analiza la variable `avg_daily_sm_hours` para identificar
en qué intervalos se concentran los registros y cuánto varía
el tiempo de uso dentro del conjunto de datos sintético.

```python
horas_redes = df["avg_daily_sm_hours"]

print(f"Registros: {horas_redes.count()}")
print(f"Promedio: {horas_redes.mean():.2f} horas")
print(f"Mediana: {horas_redes.median():.2f} horas")
print(f"Mínimo: {horas_redes.min():.2f} horas")
print(f"Máximo: {horas_redes.max():.2f} horas")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.hist(
    horas_redes,
    bins=9,
    color="steelblue",
    edgecolor="white"
)

plt.axvline(
    horas_redes.mean(),
    color="darkorange",
    linestyle="--",
    label=f"Promedio: {horas_redes.mean():.2f} h"
)

plt.axvline(
    horas_redes.median(),
    color="darkred",
    linestyle=":",
    label=f"Mediana: {horas_redes.median():.2f} h"
)

plt.title("Distribución del uso diario de redes sociales — datos sintéticos")
plt.xlabel("Tiempo diario de uso de redes sociales (horas)")
plt.ylabel("Número de registros")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
horas_validas = df["avg_daily_sm_hours"].dropna()

cantidad_mayor_3 = (horas_validas > 3).sum()
porcentaje_mayor_3 = cantidad_mayor_3 / len(horas_validas) * 100

print(f"Registros con más de 3 horas: {cantidad_mayor_3}")
print(f"Porcentaje: {porcentaje_mayor_3:.2f}%")
```

En el conjunto de datos sintético, el tiempo diario de uso de redes sociales presenta un promedio de 4,03 horas y una mediana de 3,50 horas. El intervalo con mayor frecuencia se encuentra aproximadamente entre 1,4 y 2,4 horas. La distribución se extiende hacia valores altos de uso, por lo que el promedio supera a la mediana. Estos resultados muestran distintos niveles de uso dentro del conjunto, pero no permiten establecer todavía una relación con la productividad.

El 59.67% de los registros supera las 3 horas diarias de uso de redes sociales, equivalente a 179 de 300 registros.

## 2. ¿Cómo varía el tiempo promedio de uso de redes sociales según el grupo de edad? 

```python

registros_por_edad = df.groupby("age_group").size()
print(registros_por_edad)
```

```python
horas_por_edad = (
    df.groupby("age_group")["avg_daily_sm_hours"]
    .mean()
)

print(horas_por_edad)
```

```python
resumen_por_edad = (
    df.groupby("age_group")["avg_daily_sm_hours"]
    .agg(["count", "mean", "mean"])
    .rename(columns ={
        "count" : "registros",
        "mean": "promedio_horas",
        "median": "mediana_horas"
    })
)

display(resumen_por_edad.round(2))
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 5))

barras = ax.bar(
    resumen_por_edad.index,
    resumen_por_edad["promedio_horas"],
    color="steelblue"
)

ax.bar_label(barras, fmt="%.2f", padding=3)

ax.set_title(
    "Uso diario promedio de redes sociales por edad — datos sintéticos"
)
ax.set_xlabel("Grupo de edad (años)")
ax.set_ylabel("Uso diario promedio (horas)")
ax.set_ylim(0, resumen_por_edad["promedio_horas"].max() + 1)

plt.tight_layout()
plt.show()
```

En el conjunto de datos sintético, el uso promedio diario de redes sociales disminuye progresivamente entre los grupos de edad. Los perfiles de 13 a 17 años presentan el promedio más alto, con 7,44 horas diarias, mientras que el grupo de 60 años o más registra el menor, con 1,17 horas. La diferencia entre ambos extremos es de aproximadamente 6,27 horas diarias. Este patrón describe un mayor tiempo de uso en los perfiles jóvenes del dataset, pero no permite concluir que envejecer cause una reducción del uso ni generalizar el resultado a personas reales.

## 4. ¿Qué relación existe entre las horas de sueño y la productividad autoevaluada?

```python
#Seleccionamos los datos

columnas_sueno = [
    "avg_sleep_hours",
    "productivity_self_rating"
]

datos_sueno = df[columnas_sueno].dropna().copy()

print("Registros para el analisis:", len(datos_sueno))
print("Registros exclusivos:", len(df) - len(datos_sueno))

display(datos_sueno.head())
```

```python
#resumimos las variables

resumen_sueno = (
    datos_sueno
    .agg(["count", "mean", "median", "min", "max"])
    .T #las variables originales pasan a ser filas y las estadisticas columnas
    .rename(columns={
        "count": "registros",
        "mean": "promedio",
        "median": "mediana",
        "min": "minimo",
        "max": "maximo"
    })
)

display(resumen_sueno.round(2))
```

Mediante esta tabla identificamos los valores centrales y el rango de cada variables antes de compararla

## Creamos grafica de dispercion

```python
x = datos_sueno["avg_sleep_hours"]
y = datos_sueno["productivity_self_rating"]

fig, ax = plt.subplots(figsize=(9, 5))

ax.scatter(
    x,
    y,
    color="seagreen",
    alpha=0.6,
    s=40
)

ax.set_title(
    "Horas de sueño y productividad autoevaluada — datos sintéticos"
)
ax.set_xlabel("Horas promedio de sueño")
ax.set_ylabel("Productividad autoevaluada (puntuación)")
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

### Interpretación

En el gráfico se observa que los registros con más horas
de sueño tienden a presentar puntuaciones más altas de
productividad autoevaluada.

Visualmente, los puntos muestran una tendencia ascendente:
a medida que aumentan las horas de sueño, también tienden
a aumentar las puntuaciones de productividad.

Este patrón describe los registros del conjunto de datos
sintético y no permite concluir que dormir más cause
una mayor productividad.

## 5. ¿Cómo varían la cantidad promedio de tareas completadas y la productividad autoevaluada entre niveles de uso de redes sociales?

Se clasifican los registros en intervalos de uso diario de redes
sociales para comparar el promedio de tareas completadas
y la productividad autoevaluada.

Estos intervalos se utilizan para organizar el análisis y no
representan niveles clínicos de adicción.

```python
columnas = [
    "avg_daily_sm_hours",
    "tasks_completed_per_day",
    "productivity_self_rating"
]

datos_niveles = df[columnas].dropna().copy()

print("Registros para el análisis:", len(datos_niveles))
display(datos_niveles.head())
```

```python
niveles = [
    "Hasta 2 h",
    "Más de 2 a 4 h",
    "Más de 4 a 6 h",
    "Más de 6 h"
]

datos_niveles["nivel_uso_redes"] = pd.cut(
    datos_niveles["avg_daily_sm_hours"],
    bins=[0, 2, 4, 6, float("inf")],
    labels=niveles,
    include_lowest=True
)

display(
    datos_niveles[
        ["avg_daily_sm_hours", "nivel_uso_redes"]
    ].head(10)
)
```

```python
resumen_niveles = (
    datos_niveles
    .groupby("nivel_uso_redes", observed=True)
    .agg(
        registros=("avg_daily_sm_hours", "size"),
        promedio_tareas=("tasks_completed_per_day", "mean"),
        promedio_productividad=("productivity_self_rating", "mean")
    )
)

display(resumen_niveles.round(2))
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

barras = ax.bar(
    resumen_niveles.index.astype(str),
    resumen_niveles["promedio_tareas"],
    color="steelblue"
)

ax.bar_label(barras, fmt="%.2f", padding=3)

ax.set_title(
    "Tareas completadas por nivel de uso de redes — datos sintéticos"
)
ax.set_xlabel("Tiempo diario de uso de redes sociales")
ax.set_ylabel("Promedio de tareas completadas por día")

ax.set_ylim(
    0,
    resumen_niveles["promedio_tareas"].max() + 1
)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

barras = ax.bar(
    resumen_niveles.index.astype(str),
    resumen_niveles["promedio_productividad"],
    color="seagreen"
)

ax.bar_label(barras, fmt="%.2f", padding=3)

ax.set_title(
    "Productividad por nivel de uso de redes — datos sintéticos"
)
ax.set_xlabel("Tiempo diario de uso de redes sociales")
ax.set_ylabel("Productividad autoevaluada promedio (puntuación)")

ax.set_ylim(
    0,
    resumen_niveles["promedio_productividad"].max() + 1
)

plt.tight_layout()
plt.show()
```
